```python
#Identifying cold starts
df_cp = df_cp.sort_values(['app', 'func','timestamp'])
df_cp['time_gap'] = df_cp.groupby(['app','func'])['timestamp'].diff().dt.total_seconds()
cold_start_threshold = 600
df_cp['is_potential_coldstart'] = ((df_cp['time_gap']>cold_start_threshold) | (df_cp['time_gap'].isna())).astype(int)
```

```python
#Function STatistics Analysis

func_stats = df_cp.groupby(['app', 'func']).agg({
    'duration':['count', 'mean', 'std', 'min', 'max'],
    'is_potential_coldstart':['sum', 'mean'],
    'time_gap': ['mean'],
    'timestamp':['min','max']
}).reset_index()


# Flatten column names
func_stats.columns = [
        'app', 'func', 'invocation_count', 'avg_duration', 'std_duration',
        'min_duration', 'max_duration', 'cold_starts', 'cold_start_rate',
        'avg_time_gap', 'first_invocation', 'last_invocation'
]

func_stats['activity_span_hours'] = (func_stats['last_invocation'] - func_stats['first_invocation']).dt.total_seconds()/3600
func_stats['invocations_per_hour'] = func_stats['invocation_count']/np.maximum(func_stats['activity_span_hours'],0.1)
func_stats['duration_variability'] = func_stats['std_duration']/np.maximum(func_stats['avg_duration'], 0.001)

```

```python

# Filter functions with reasonable activity (>50 invocations)
active_funcs = func_stats[func_stats['invocation_count'] >= 50].copy()
```

```python
active_funcs['regularity_score'] = 1 / (1 + active_funcs['duration_variability'])  # Higher = more regular
active_funcs['activity_intensity'] = active_funcs['invocations_per_hour']
```

```python
active_funcs.describe()
```

```python
# ----- CONDITIONS  -----

conditions = [

    # Sporadic
    (
        (active_funcs["invocations_per_hour"] < 5) &
        (active_funcs["cold_start_rate"] > 0.20) &
        (active_funcs["avg_time_gap"] > 2000)
    ),

    # Steady-High
    (
        (active_funcs["invocations_per_hour"] > 100) &
        (active_funcs["cold_start_rate"] < 0.05) &
        (active_funcs["duration_variability"] < 1.0) &
        (active_funcs["regularity_score"] > 0.60)
    ),

    # Regular
    (
        (active_funcs["invocations_per_hour"].between(10, 100)) &
        (active_funcs["regularity_score"] >= 0.58) &
        (active_funcs["duration_variability"] < 1.5)
    ),

    # Bursty
    (
        (active_funcs["duration_variability"] > 2.0) |
        ((active_funcs["invocations_per_hour"] > 100) & (active_funcs["cold_start_rate"] > 0.20)) |
        (active_funcs["invocation_count"] > np.percentile(active_funcs["invocation_count"], 90))
    ),
]

choices = [
    "Sporadic",
    "Steady-High",
    "Regular",
    "Bursty"
]
```

```python

active_funcs['workload_type'] = np.select(conditions, choices, default='Mixed')

workload_summary = active_funcs['workload_type'].value_counts()
print(f"Workload type distribution (functions with ≥50 invocations):")
for workload_type, count in workload_summary.items():
  percentage = (count / len(active_funcs)) * 100
  print(f"  {workload_type}: {count} functions ({percentage:.1f}%)")
```

```python
df_features = df_cp.sort_values(['app', 'func', 'timestamp'])

print("Creating cyclical time features...")
# Cyclical encoding for time features (better for ML models)
df_features['hour_sin'] = np.sin(2 * np.pi * df_features['hour'] / 24)
df_features['hour_cos'] = np.cos(2 * np.pi * df_features['hour'] / 24)
df_features['dayofweek_sin'] = np.sin(2 * np.pi * df_features['dayofweek'] / 7)
df_features['dayofweek_cos'] = np.cos(2 * np.pi * df_features['dayofweek'] / 7)
df_features['dayofmonth_sin'] = np.sin(2 * np.pi * df_features['dayofmonth'] / 31)
df_features['dayofmonth_cos'] = np.cos(2 * np.pi * df_features['dayofmonth'] / 31)
```

```python
# Business hours and weekend patterns
df_features['is_business_hours'] = ((df_features['hour'] >= 9) &
                                      (df_features['hour'] <= 17) &
                                       (df_features['is_weekend'] == 0)).astype(int)

```

```python
df_features['is_peak_hours'] = ((df_features['hour'] >= 10) & (df_features['hour'] <= 15)).astype(int)
```

```python
df_features['is_evening'] = ((df_features['hour'] >= 18) &
                                (df_features['hour'] <= 23)).astype(int)
```

```python
df_features = df_features.sort_values(by=['app', 'func', 'timestamp'])

# Calculate Rolling Features

windows = [5, 15, 30]
for window in windows:
  # Rolling mean of durations
  df_features[f'duration_rolling_mean_{window}'] = (
      df_features.groupby(['app', 'func'])['duration']
      .rolling(window=window, min_periods=1)
      .mean()
      .reset_index(level=[0, 1], drop=True) #to drop the new app and func levels created by groupby
  )

  # Rolling std of durations
  df_features[f'duration_rolling_std_{window}'] = (
      df_features.groupby(['app', 'func'])['duration']
      .rolling(window=window, min_periods=1)
      .std()
      .fillna(0)
      .reset_index(level=[0, 1], drop=True)
  )

  # Rolling count (frequency)
  df_features[f'frequency_{window}'] = (
      df_features.groupby(['app', 'func'])['duration']
      .rolling(window=window, min_periods=1)
      .count()
      .reset_index(level=[0, 1], drop=True)
  )

print("Creating lag features...")
#Lag features
lags = [1, 2, 3, 5]
for lag in lags:
    df_features[f'duration_lag_{lag}'] = (
        df_features.groupby(['app', 'func'])['duration']
        .shift(lag)
        .fillna(df_features['duration'].mean())
    )
```

```python
# Inter-arrival time features
df_features['time_since_last'] = df_features.groupby(['app', 'func'])['timestamp'].diff().dt.total_seconds().fillna(0)
df_features['time_since_last_log'] = np.log1p(df_features['time_since_last'])

# Inverse frequency (how rare are invocations for this function)
df_features['inv_frequency'] = 1.0 / (df_features.groupby(['app', 'func']).cumcount() + 1)

print("Creating burst detection features...")
# Burst detection
df_features['is_burst'] = (df_features['time_since_last'] < 10).astype(int)  # Calls within 10 seconds
df_features['burst_size'] = df_features.groupby(['app', 'func'])['is_burst'].rolling(10, min_periods=1).sum().reset_index(level=[0, 1], drop=True)
```

```python
def calculate_mape(y_true, y_pred):
    #Calculate MAPE, handling zero values
    mask = y_true != 0
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100
```

```python
def load_and_preprocess(csv_path, func_id):
    #Load and preprocess time series data
    try:
        df = pd.read_csv(csv_path)
        df['ds'] = pd.to_datetime(df['ds'])
        df['y'] = pd.to_numeric(df['y'], errors='coerce')
        df = df.dropna(subset=['ds', 'y'])

        # Floor to minute precision
        df['ds'] = df['ds'].dt.floor('1min')

        # Add small offset to avoid zero issues
        df['y'] = df['y'] + 0.1

        return df
    except Exception as e:
        print(f"Error loading {csv_path}: {e}")
        return None
```

```python
def train_prophet(df, func_id, output_dir='models/prophet'):
    #Train Prophet model with logistic growth
    print(f"\n")
    print(f"Training Prophet for {func_id}")


    df = df.copy()

    # Train/test split
    train_size = int(len(df) * TRAIN_RATIO)
    train_df = df[:train_size]
    test_df = df[train_size:]

    print(f"Total points: {len(df):,}")
    print(f"Train: {len(train_df):,} | Test: {len(test_df):,}")

    # Set cap for logistic growth
    cap_value = train_df['y'].quantile(0.99)
    train_df['cap'] = cap_value
    train_df['floor'] = 0.0

    print(f"Growth cap: {cap_value:.2f}")

    # Build Prophet model
    model = Prophet(
        growth='logistic',
        seasonality_mode='multiplicative',
        changepoint_prior_scale=0.05,
        seasonality_prior_scale=10.0,
        daily_seasonality=True,
        weekly_seasonality=True,
        yearly_seasonality=False,
        interval_width=0.95
    )

    model.add_seasonality(name='intra_day', period=1.0, fourier_order=10)
    model.add_seasonality(name='intra_hour', period=1/24.0, fourier_order=5)

    # Train
    print("Training Prophet...")
    model.fit(train_df[['ds', 'y', 'cap', 'floor']])

    # Predict on test set
    future = test_df[['ds']].copy()
    future['cap'] = cap_value
    future['floor'] = 0.0

    forecast = model.predict(future)

    # Adjust predictions (remove offset)
    prophet_pred = np.maximum(forecast['yhat'].values - 0.1, 0)
    actual = np.maximum(test_df['y'].values - 0.1, 0)

    # Calculate metrics
    mae = mean_absolute_error(actual, prophet_pred)
    rmse = np.sqrt(mean_squared_error(actual, prophet_pred))
    mape = calculate_mape(actual, prophet_pred)

    print(f"\nProphet Performance:")
    print(f"   MAE:  {mae:.2f}")
    print(f"   RMSE: {rmse:.2f}")
    print(f"   MAPE: {mape:.2f}%")

    # Save model
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    model_file = output_path / f'{func_id}_prophet.pkl'
    joblib.dump(model, model_file)
    print(f"Saved: {model_file}")

    return {
        'function': func_id,
        'model': model,
        'mae': mae,
        'rmse': rmse,
        'mape': mape,
        'train_df': train_df,
        'test_df': test_df,
        'forecast': forecast,
        'prophet_pred': prophet_pred,
        'actual': actual,
        'cap_value': cap_value
    }
```

LSTM Model

```python
#Hyperparameters
LSTM_LOOKBACK = 30
LSTM_UNITS = 50
LSTM_LAYERS = 2
DROPOUT_RATE = 0.2
EPOCHS = 100
BATCH_SIZE = 32
VALIDATION_SPLIT = 0.2
```

```python
def create_sequences(data, lookback):
    #Create sequences for LSTM training
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:i + lookback])
        y.append(data[i + lookback])
    return np.array(X), np.array(y)
```

```python
def train_lstm_only(df, func_id, output_dir='models/lstm'):
    #Train LSTM-only model
    print(f"\n")
    print(f"Training LSTM-only for {func_id}")


    df = df.copy()

    # Train/test split
    train_size = int(len(df) * TRAIN_RATIO)
    train_df = df[:train_size]
    test_df = df[train_size:]

    print(f"Total points: {len(df):,}")
    print(f"Train: {len(train_df):,} | Test: {len(test_df):,}")

    # Extract values
    train_data = train_df['y'].values
    test_data = test_df['y'].values

    # Create sequences
    print(f"Creating sequences (lookback={LSTM_LOOKBACK})...")
    X_train, y_train = create_sequences(train_data, LSTM_LOOKBACK)

    if len(X_train) < 50:
        print(f"   Insufficient data: {len(X_train)} samples")
        return None

    print(f"Training sequences: {len(X_train):,}")

    # Split into train/validation
    split_idx = int(len(X_train) * (1 - VALIDATION_SPLIT))
    X_train_final = X_train[:split_idx]
    X_val = X_train[split_idx:]
    y_train_final = y_train[:split_idx]
    y_val = y_train[split_idx:]

    # Scale data
    X_scaler = MinMaxScaler()
    y_scaler = MinMaxScaler()

    X_train_scaled = X_scaler.fit_transform(
        X_train_final.reshape(-1, 1)
    ).reshape(X_train_final.shape[0], LSTM_LOOKBACK, 1)

    X_val_scaled = X_scaler.transform(
        X_val.reshape(-1, 1)
    ).reshape(X_val.shape[0], LSTM_LOOKBACK, 1)

    y_train_scaled = y_scaler.fit_transform(y_train_final.reshape(-1, 1))
    y_val_scaled = y_scaler.transform(y_val.reshape(-1, 1))

    # Build LSTM model
    print(f"Building LSTM ({LSTM_LAYERS} layers, {LSTM_UNITS} units)...")

    model = Sequential()
    model.add(LSTM(
        units=LSTM_UNITS,
        return_sequences=(LSTM_LAYERS > 1),
        input_shape=(LSTM_LOOKBACK, 1)
    ))
    model.add(Dropout(DROPOUT_RATE))

    for i in range(1, LSTM_LAYERS):
        model.add(LSTM(
            units=LSTM_UNITS,
            return_sequences=(i < LSTM_LAYERS - 1)
        ))
        model.add(Dropout(DROPOUT_RATE))

    model.add(Dense(1))
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])

    # Train
    print("Training LSTM...")
    early_stopping = EarlyStopping(
        monitor='val_loss',
        patience=10,
        restore_best_weights=True,
        verbose=0
    )

    history = model.fit(
        X_train_scaled, y_train_scaled,
        validation_data=(X_val_scaled, y_val_scaled),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[early_stopping],
        verbose=0
    )

    print(f"Trained in {len(history.history['loss'])} epochs")
    print(f"   Final val loss: {history.history['val_loss'][-1]:.4f}")

    # Test predictions
    print(" Evaluating on test set...")


    test_data_with_context = np.concatenate([train_data[-LSTM_LOOKBACK:], test_data])
    X_test, y_test = create_sequences(test_data_with_context, LSTM_LOOKBACK)

    # Match test set length
    X_test = X_test[:len(test_data)]
    y_test = y_test[:len(test_data)]

    # Scale and predict
    X_test_scaled = X_scaler.transform(
        X_test.reshape(-1, 1)
    ).reshape(X_test.shape[0], LSTM_LOOKBACK, 1)

    y_pred_scaled = model.predict(X_test_scaled, verbose=0)
    y_pred = y_scaler.inverse_transform(y_pred_scaled).flatten()

    # Remove offset
    y_pred = np.maximum(y_pred - 0.1, 0)
    y_test_adjusted = np.maximum(y_test - 0.1, 0)

    # Calculate metrics
    mae = mean_absolute_error(y_test_adjusted, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test_adjusted, y_pred))
    mape = calculate_mape(y_test_adjusted, y_pred)

    print(f"\n LSTM-Only Performance:")
    print(f"   MAE:  {mae:.2f}")
    print(f"   RMSE: {rmse:.2f}")
    print(f"   MAPE: {mape:.2f}%")

    # Save model and scalers
    output_path = Path(output_dir)
    output_path.mkdir(parents=True, exist_ok=True)
    model_file = output_path / f'{func_id}_lstm.h5'
    scaler_file = output_path / f'{func_id}_lstm_scalers.pkl'

    # Save model
    model.save(model_file, save_format='h5')

    # Save scalers
    scalers = {
        'X_scaler': X_scaler,
        'y_scaler': y_scaler
    }
    joblib.dump(scalers, scaler_file)

    print(f"Saved:")
    print(f"   Model: {model_file}")
    print(f"   Scalers: {scaler_file}")

    return {
        'function': func_id,
        'model': model,
        'X_scaler': X_scaler,
        'y_scaler': y_scaler,
        'mae': mae,
        'rmse': rmse,
        'mape': mape,
        'y_test': y_test_adjusted,
        'y_pred': y_pred,
        'test_dates': test_df['ds'].values[:len(y_pred)]
    }
```

```python
LSTM_LOOKBACK = 30
LSTM_UNITS = 50
LSTM_LAYERS = 2
DROPOUT_RATE = 0.2
EPOCHS = 100
BATCH_SIZE = 32
VALIDATION_SPLIT = 0.2
TRAIN_RATIO = 0.8
OUTPUT_DIR = 'models/hybrid'
```

```python
#Train Hybrid models (Prophet + LSTM on residuals)


FUNCTION_IDS = [
    'func_235', 'func_242', 'func_126', 'func_186',
    'func_190', 'func_110', 'func_99', 'func_191'
]




def create_sequences(data, lookback):
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:i + lookback])
        y.append(data[i + lookback])
    return np.array(X), np.array(y)


def calculate_mape(y_true, y_pred):
    mask = y_true != 0
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100


def load_and_preprocess(csv_path):
    try:
        df = pd.read_csv(csv_path)
        df['ds'] = pd.to_datetime(df['ds'])
        df['y'] = pd.to_numeric(df['y'], errors='coerce')
        df = df.dropna(subset=['ds', 'y'])
        df['ds'] = df['ds'].dt.floor('1min')
        df['y'] = df['y'] + 0.1
        return df
    except Exception as e:
        print(f" Error loading {csv_path}: {e}")
        return None


def train_hybrid_single(func_id):
    #Train Hybrid model for a single function
    print(f"\n")
    print(f"Training Hybrid for {func_id}")


    # Load data
    csv_path = f'timeseries_{func_id}.csv'
    df = load_and_preprocess(csv_path)

    if df is None:
        return None

    # Train/test split
    train_size = int(len(df) * TRAIN_RATIO)
    train_df = df[:train_size].copy()
    test_df = df[train_size:].copy()

    print(f"Total points: {len(df):,}")
    print(f"Train: {len(train_df):,} | Test: {len(test_df):,}")

    #  Train Prophet ==========
    print("\n[1/3] Training Prophet component...")

    cap_value = train_df['y'].quantile(0.99)
    train_df['cap'] = cap_value
    train_df['floor'] = 0.0

    prophet_model = Prophet(
        growth='logistic',
        seasonality_mode='multiplicative',
        changepoint_prior_scale=0.05,
        seasonality_prior_scale=10.0,
        daily_seasonality=True,
        weekly_seasonality=True,
        yearly_seasonality=False,
        interval_width=0.95
    )

    prophet_model.add_seasonality(name='intra_day', period=1.0, fourier_order=10)
    prophet_model.add_seasonality(name='intra_hour', period=1/24.0, fourier_order=5)

    prophet_model.fit(train_df[['ds', 'y', 'cap', 'floor']])

    # Get Prophet predictions on test set
    future = test_df[['ds']].copy()
    future['cap'] = cap_value
    future['floor'] = 0.0

    forecast = prophet_model.predict(future)
    prophet_pred = np.maximum(forecast['yhat'].values - 0.1, 0)
    actual = np.maximum(test_df['y'].values - 0.1, 0)

    #  Calculate Residuals ==========
    print("\n[2/3] Calculating Prophet residuals...")

    residuals = actual - prophet_pred

    print(f"   Mean: {residuals.mean():.2f}")
    print(f"   Std:  {residuals.std():.2f}")
    print(f"   Range: [{residuals.min():.2f}, {residuals.max():.2f}]")

    if len(residuals) < LSTM_LOOKBACK + 50:
        print(f" Insufficient data for LSTM: {len(residuals)} points")
        print("  Saving Prophet-only model")

        output_path = Path(OUTPUT_DIR)
        output_path.mkdir(parents=True, exist_ok=True)
        prophet_file = output_path / f'{func_id}_prophet.pkl'
        joblib.dump(prophet_model, prophet_file)

        prophet_mae = mean_absolute_error(actual, prophet_pred)

        return {
            'function': func_id,
            'prophet_mae': prophet_mae,
            'hybrid_mae': prophet_mae,
            'improvement': 0.0,
            'insufficient_data': True
        }

    #  Train LSTM on Residuals ==========
    print("\n[3/3] Training LSTM on residuals...")

    X, y = create_sequences(residuals, LSTM_LOOKBACK)
    print(f"   LSTM sequences: {len(X):,}")

    # Split
    split_idx = int(len(X) * (1 - VALIDATION_SPLIT))
    X_train, X_val = X[:split_idx], X[split_idx:]
    y_train, y_val = y[:split_idx], y[split_idx:]

    # Scale
    X_scaler = MinMaxScaler()
    y_scaler = MinMaxScaler()

    X_train_scaled = X_scaler.fit_transform(X_train.reshape(-1, 1)).reshape(
        X_train.shape[0], LSTM_LOOKBACK, 1)
    X_val_scaled = X_scaler.transform(X_val.reshape(-1, 1)).reshape(
        X_val.shape[0], LSTM_LOOKBACK, 1)

    y_train_scaled = y_scaler.fit_transform(y_train.reshape(-1, 1))
    y_val_scaled = y_scaler.transform(y_val.reshape(-1, 1))

    # Build LSTM
    lstm_model = Sequential()
    lstm_model.add(LSTM(
        units=LSTM_UNITS,
        return_sequences=(LSTM_LAYERS > 1),
        input_shape=(LSTM_LOOKBACK, 1)
    ))
    lstm_model.add(Dropout(DROPOUT_RATE))

    for i in range(1, LSTM_LAYERS):
        lstm_model.add(LSTM(
            units=LSTM_UNITS,
            return_sequences=(i < LSTM_LAYERS - 1)
        ))
        lstm_model.add(Dropout(DROPOUT_RATE))

    lstm_model.add(Dense(1))
    lstm_model.compile(optimizer='adam', loss='mse', metrics=['mae'])

    # Train
    early_stopping = EarlyStopping(
        monitor='val_loss',
        patience=10,
        restore_best_weights=True,
        verbose=0
    )

    history = lstm_model.fit(
        X_train_scaled, y_train_scaled,
        validation_data=(X_val_scaled, y_val_scaled),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[early_stopping],
        verbose=0
    )

    print(f"Trained in {len(history.history['loss'])} epochs")

    # Evaluate Hybrid ==========
    print("\nEvaluating Hybrid model...")

    X_full, _ = create_sequences(residuals, LSTM_LOOKBACK)
    X_full_scaled = X_scaler.transform(X_full.reshape(-1, 1)).reshape(
        X_full.shape[0], LSTM_LOOKBACK, 1)

    lstm_residual_pred_scaled = lstm_model.predict(X_full_scaled, verbose=0)
    lstm_residual_pred = y_scaler.inverse_transform(lstm_residual_pred_scaled).flatten()

    # Align predictions
    prophet_pred_aligned = prophet_pred[LSTM_LOOKBACK:]
    actual_aligned = actual[LSTM_LOOKBACK:]

    # Create hybrid predictions
    hybrid_pred = prophet_pred_aligned + lstm_residual_pred
    hybrid_pred = np.maximum(hybrid_pred, 0)

    # Calculate metrics
    prophet_mae = mean_absolute_error(actual_aligned, prophet_pred_aligned)
    hybrid_mae = mean_absolute_error(actual_aligned, hybrid_pred)
    hybrid_rmse = np.sqrt(mean_squared_error(actual_aligned, hybrid_pred))
    hybrid_mape = calculate_mape(actual_aligned, hybrid_pred)

    improvement = ((prophet_mae - hybrid_mae) / prophet_mae) * 100

    print(f"\nResults:")
    print(f"   Prophet MAE: {prophet_mae:.2f}")
    print(f"   Hybrid MAE:  {hybrid_mae:.2f}")
    print(f"   Hybrid RMSE: {hybrid_rmse:.2f}")
    print(f"   Improvement: {improvement:+.2f}%")

    if improvement > 0:
        print(f"Hybrid is better!")
    else:
        print(f" Prophet alone is better")

    # Save Models ==========
    output_path = Path(OUTPUT_DIR)
    output_path.mkdir(parents=True, exist_ok=True)

    prophet_file = output_path / f'{func_id}_prophet.pkl'
    lstm_file = output_path / f'{func_id}_lstm_residual.h5'
    scaler_file = output_path / f'{func_id}_lstm_scalers.pkl'

    # Save Prophet
    joblib.dump(prophet_model, prophet_file)

    lstm_model.save(str(lstm_file))

    # Save scalers
    scalers = {'X_scaler': X_scaler, 'y_scaler': y_scaler}
    joblib.dump(scalers, scaler_file)

    print(f"Saved:")
    print(f"   Prophet: {prophet_file}")
    print(f"   LSTM: {lstm_file}")
    print(f"   Scalers: {scaler_file}")

    return {
        'function': func_id,
        'prophet_mae': prophet_mae,
        'hybrid_mae': hybrid_mae,
        'hybrid_rmse': hybrid_rmse,
        'hybrid_mape': hybrid_mape,
        'improvement': improvement,
        'insufficient_data': False,
        "test_dates": test_df['ds'].iloc[LSTM_LOOKBACK:].values,
        "actual_aligned": actual_aligned,
        "prophet_pred_aligned": prophet_pred_aligned,
        "hybrid_pred": hybrid_pred
    }


def train_all_hybrid():
    #Train Hybrid for all functions
    print("\n")
    print("BATCH HYBRID TRAINING")


    hybrid_results = []

    for func_id in tqdm(FUNCTION_IDS, desc="Training Hybrid models"):
        try:
            result = train_hybrid_single(func_id)
            if result:
                hybrid_results.append(result)
        except Exception as e:
            print(f" Error training {func_id}: {e}")
            import traceback
            traceback.print_exc()

    # Summary
    if hybrid_results:
        print("\n")
        print("HYBRID TRAINING SUMMARY")


        results_df = pd.DataFrame(hybrid_results)
        print("\n" + results_df.to_string(index=False))

        print(f"\n Average Performance:")
        print(f"   Prophet MAE: {results_df['prophet_mae'].mean():.2f}")
        print(f"   Hybrid MAE:  {results_df['hybrid_mae'].mean():.2f}")
        print(f"   Avg Improvement: {results_df['improvement'].mean():+.2f}%")

        improved = (results_df['improvement'] > 0).sum()
        print(f"\n Functions where Hybrid improved: {improved}/{len(results_df)}")

        results_df.to_csv('hybrid_training_results.csv', index=False)
        print(f"\n Results saved to: hybrid_training_results.csv")

    return hybrid_results


if __name__ == "__main__":
    hybrid_results = train_all_hybrid()
    print("\nHybrid training complete!")
```

```python
def visualize_hybrid(result):
    dates = result["test_dates"]
    actual = result["actual_aligned"]
    prophet = result["prophet_pred_aligned"]
    hybrid = result["hybrid_pred"]

    plt.figure(figsize=(14,5))
    plt.plot(dates, actual, label="Actual")
    plt.plot(dates, prophet, label="Prophet")
    plt.plot(dates, hybrid, label="Hybrid")
    plt.title(f"Hybrid – Actual vs Prophet vs Hybrid ({result['function']})")
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()
```

```python
def save_all_models(results, hybrid_results, lstm_only_results):
    import os, json, pickle, joblib, shutil
    from google.colab import files

    print("=== SAVING MODELS ")

    # ---- PROPHET MODELS ----
    print("\n[1] Checking Prophet models...")
    for item in results:
        func_id = item["function"]
        pkl_path = f"models/prophet/{func_id}_prophet.pkl"

        if os.path.exists(pkl_path):
            print(f"[Prophet] OK -> {pkl_path}")
        else:
            print(f"[Prophet] MISSING -> {pkl_path}")

    # ---- HYBRID MODELS ----
    print("\n[2] Checking Hybrid models...")
    for item in hybrid_results:
        func_id = item["function"]

        prophet_path = f"models/hybrid/{func_id}_prophet.pkl"
        lstm_path = f"models/hybrid/{func_id}_lstm_residual.h5"
        scaler_path = f"models/hybrid/{func_id}_lstm_scalers.pkl"

        print(f"\nHybrid {func_id}:")
        print(f"  Prophet: {'OK' if os.path.exists(prophet_path) else 'MISSING'}")
        print(f"  LSTM: {'OK' if os.path.exists(lstm_path) else 'MISSING'}")
        print(f"  Scalers: {'OK' if os.path.exists(scaler_path) else 'MISSING'}")

    # ---- LSTM-ONLY ----
    print("\n[3] Checking LSTM-only models...")
    for item in lstm_only_results:
        func_id = item["function"]

        lstm_path = f"models/lstm/{func_id}_lstm.h5"
        scaler_path = f"models/lstm/{func_id}_lstm_scalers.pkl"

        print(f"\nLSTM {func_id}:")
        print(f"  Model: {'OK' if os.path.exists(lstm_path) else 'MISSING'}")
        print(f"  Scalers: {'OK' if os.path.exists(scaler_path) else 'MISSING'}")

    # ---- METADATA ----
    metadata = {
        "functions": FUNCTION_IDS,
        "models_saved": {
            "prophet": len(os.listdir("models/prophet")),
            "hybrid": len(os.listdir("models/hybrid")),
            "lstm": len(os.listdir("models/lstm"))
        }
    }

    with open("models/metadata.json", "w") as f:
        json.dump(metadata, f, indent=2)

    print("\nMetadata written.")

    # ---- ZIP MODELS ----
    shutil.make_archive("models", "zip", "models")
    files.download("models.zip")

    print("\nModels packaged and downloaded!")
```
